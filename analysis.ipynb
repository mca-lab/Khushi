```python
import pandas as pd

pokemon = pd.read_csv("data/raw/pokemon.csv")
pokemon_species = pd.read_csv("data/raw/pokemon_species.csv")

pokemon.head(), pokemon_species.head()
```

```python
pokemon.head()
```

```python
pokemon_species.head()
```

```python
merged = pd.merge(
    pokemon,
    pokemon_species,
    on="id",
    how="inner"
)

merged.head()
```

```python
merged.head()
```

```python
pokemon.columns
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

stats = ["height", "weight", "base_experience", "order"]

for stat in stats:
    plt.figure(figsize=(8,4))
    sns.histplot(pokemon[stat], bins=30, kde=True)
    plt.title(f"Distribution of {stat.capitalize()}")
    plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=merged,
    x="height",
    y="weight",
    hue="is_legendary",
    palette="coolwarm"
)
plt.title("Height vs Weight (Legendary vs Non-Legendary)")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=merged, x="habitat_id", y="base_experience", estimator="mean")
plt.title("Average Base Experience by Habitat")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.regplot(data=merged, x="capture_rate", y="base_experience")
plt.title("Capture Rate vs Base Experience")
plt.show()
```

```python
plt.figure(figsize=(12,8))
numeric = merged.select_dtypes(include=['int64','float64'])
sns.heatmap(numeric.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap – Merged Dataset")
plt.show()
```

```python
sns.boxplot(data=merged, x="is_legendary", y="base_happiness")
plt.title("Base Happiness: Legendary vs Non-Legendary Pokémon")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(data=merged, x="shape_id", hue="generation_id")
plt.title("Distribution of Pokémon Shapes by Generation")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.violinplot(data=merged, x="habitat_id", y="capture_rate")
plt.title("Capture Rate Distribution Across Habitats")
plt.show()
```
